```python
from dateutil import parser
date = parser.parse("4th of July, 2021")
date
```

```python
# %A: 요일 전체이름 = Monday
# %a: 요일 이름 = Mon
date.strftime('%A')
date.strftime('%a')
```

```python
import pandas as pd
date =pd.to_datetime("4th of July, 2021")
date
```

```python
import numpy as np
# day 단위의 시간 간격으로 12간의 날짜 생성
date + pd.to_timedelta(np.arange(12),'D')
```
