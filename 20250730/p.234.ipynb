```python
import yfinance as yf
ticker = 'TSLA'

df = yf.download(ticker, start='2023-01-01', end='2025-07-29')
df.head()
```

```python
df.loc[:, ('Close', 'TSLA')].plot()
```

```python
tesla_df = df['Close']
```

```python
tesla_df.plot(alpha=0.5, style='-')
```

```python
tesla_df.resample('BA').mean().plot(style=":")
```
