```python
# apply() => 시리즈(열)의 각 값에 함수를 하나하나 적용하는 함수
df['sex'].apply(lambda x: 0 if x == 'male' else 1)


def f(x):
    if x == 'male' : return 0
    return 1

df['sex'].apply(f)
df.loc[:, 'sex'] = df['sex'].apply(f)

```

```python
df.info()
```

```python
# type 변경 => loc로하면 항상 초기화됨 (주의)
df['sex'] = df['sex'].astype('int')

```
